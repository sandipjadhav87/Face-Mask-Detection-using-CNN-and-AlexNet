```python
# ============================================================
# EXPERIMENT 5
# FACE MASK DETECTION USING CNN AND ALEXNET
# ============================================================

# ============================================================
# STEP 1: INSTALL AND IMPORT LIBRARIES
# ============================================================

!pip install -q kagglehub

import os
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
import tensorflow as tf
import kagglehub

from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.metrics import classification_report, confusion_matrix

print("TensorFlow Version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices("GPU"))


# ============================================================
# STEP 2: DOWNLOAD DATASET USING KAGGLEHUB
# ============================================================

path = kagglehub.dataset_download(
    "omkargurav/face-mask-dataset"
)

print("\nDataset downloaded successfully!")
print("Dataset path:", path)


# ============================================================
# STEP 3: SET DATASET PATH
# ============================================================

# Dataset structure:
# path/
# └── data/
#     ├── with_mask/
#     └── without_mask/

data_dir = os.path.join(path, "data")

with_mask_dir = os.path.join(
    data_dir,
    "with_mask"
)

without_mask_dir = os.path.join(
    data_dir,
    "without_mask"
)

print("\nDataset folders:")
print(os.listdir(data_dir))

print("\nNumber of images:")
print("With Mask     :", len(os.listdir(with_mask_dir)))
print("Without Mask  :", len(os.listdir(without_mask_dir)))

total_images = (
    len(os.listdir(with_mask_dir))
    + len(os.listdir(without_mask_dir))
)

print("Total Images  :", total_images)


# ============================================================
# STEP 4: DISPLAY SAMPLE IMAGES
# ============================================================

fig, axes = plt.subplots(
    2, 4,
    figsize=(12, 6)
)

# With mask images
for i, fname in enumerate(
    os.listdir(with_mask_dir)[:4]
):

    img = plt.imread(
        os.path.join(with_mask_dir, fname)
    )

    axes[0, i].imshow(img)
    axes[0, i].axis("off")
    axes[0, i].set_title("With Mask")


# Without mask images
for i, fname in enumerate(
    os.listdir(without_mask_dir)[:4]
):

    img = plt.imread(
        os.path.join(without_mask_dir, fname)
    )

    axes[1, i].imshow(img)
    axes[1, i].axis("off")
    axes[1, i].set_title("Without Mask")


plt.tight_layout()
plt.show()


# ============================================================
# STEP 5: DATA PREPROCESSING AND AUGMENTATION
# ============================================================

IMG_SIZE = 128
BATCH_SIZE = 32

# Training data augmentation
train_datagen = ImageDataGenerator(
    rescale=1.0 / 255,
    rotation_range=15,
    zoom_range=0.15,
    horizontal_flip=True,
    validation_split=0.20
)

# Training generator
train_gen = train_datagen.flow_from_directory(
    data_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="binary",
    subset="training",
    shuffle=True
)

# Validation generator
# Only normalization, NO augmentation
val_datagen = ImageDataGenerator(
    rescale=1.0 / 255,
    validation_split=0.20
)

val_gen = val_datagen.flow_from_directory(
    data_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="binary",
    subset="validation",
    shuffle=False
)

print("\nClass Indices:")
print(train_gen.class_indices)


# ============================================================
# STEP 6: BUILD BASIC CNN
# ============================================================

def build_basic_cnn(
    input_shape=(128, 128, 3)
):

    model = models.Sequential(
        [

            layers.Input(
                shape=input_shape
            ),

            # Convolution Block 1
            layers.Conv2D(
                32,
                (3, 3),
                activation="relu"
            ),

            layers.MaxPooling2D(
                (2, 2)
            ),

            # Convolution Block 2
            layers.Conv2D(
                64,
                (3, 3),
                activation="relu"
            ),

            layers.MaxPooling2D(
                (2, 2)
            ),

            # Convolution Block 3
            layers.Conv2D(
                128,
                (3, 3),
                activation="relu"
            ),

            layers.MaxPooling2D(
                (2, 2)
            ),

            # Fully Connected Layers
            layers.Flatten(),

            layers.Dense(
                128,
                activation="relu"
            ),

            layers.Dropout(0.5),

            # Binary classification
            layers.Dense(
                1,
                activation="sigmoid"
            )
        ],
        name="Basic_CNN"
    )

    return model


basic_cnn = build_basic_cnn()

print("\n================ BASIC CNN ================")
basic_cnn.summary()


# ============================================================
# STEP 7: BUILD ALEXNET-INSPIRED CNN
# ============================================================

def build_alexnet(
    input_shape=(128, 128, 3)
):

    model = models.Sequential(
        [

            layers.Input(
                shape=input_shape
            ),

            # -------------------------
            # Convolution Block 1
            # -------------------------

            layers.Conv2D(
                96,
                (11, 11),
                strides=4,
                activation="relu"
            ),

            layers.MaxPooling2D(
                (3, 3),
                strides=2
            ),

            layers.BatchNormalization(),


            # -------------------------
            # Convolution Block 2
            # -------------------------

            layers.Conv2D(
                256,
                (5, 5),
                padding="same",
                activation="relu"
            ),

            layers.MaxPooling2D(
                (3, 3),
                strides=2
            ),

            layers.BatchNormalization(),


            # -------------------------
            # Convolution Blocks 3-5
            # -------------------------

            layers.Conv2D(
                384,
                (3, 3),
                padding="same",
                activation="relu"
            ),

            layers.Conv2D(
                384,
                (3, 3),
                padding="same",
                activation="relu"
            ),

            layers.Conv2D(
                256,
                (3, 3),
                padding="same",
                activation="relu"
            ),

            layers.MaxPooling2D(
                (3, 3),
                strides=2
            ),


            # -------------------------
            # Fully Connected Layers
            # -------------------------

            layers.Flatten(),

            layers.Dense(
                4096,
                activation="relu"
            ),

            layers.Dropout(0.5),

            layers.Dense(
                4096,
                activation="relu"
            ),

            layers.Dropout(0.5),

            # Binary classification
            layers.Dense(
                1,
                activation="sigmoid"
            )
        ],
        name="AlexNet"
    )

    return model


alexnet = build_alexnet()

print("\n================ ALEXNET ================")
alexnet.summary()


# ============================================================
# STEP 8: COMPILE BOTH MODELS
# ============================================================

basic_cnn.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

alexnet.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("\nBoth models compiled successfully.")


# ============================================================
# STEP 9: TRAIN BASIC CNN
# ============================================================

print("\n==========================================")
print("TRAINING BASIC CNN")
print("==========================================")

start_basic = time.time()

history_basic = basic_cnn.fit(
    train_gen,
    validation_data=val_gen,
    epochs=10
)

time_basic = time.time() - start_basic

print(
    "\nBasic CNN Training Time: "
    f"{time_basic:.2f} seconds"
)


# ============================================================
# STEP 10: TRAIN ALEXNET
# ============================================================

print("\n==========================================")
print("TRAINING ALEXNET")
print("==========================================")

start_alexnet = time.time()

history_alexnet = alexnet.fit(
    train_gen,
    validation_data=val_gen,
    epochs=10
)

time_alexnet = time.time() - start_alexnet

print(
    "\nAlexNet Training Time: "
    f"{time_alexnet:.2f} seconds"
)


# ============================================================
# STEP 11: TRAINING ACCURACY COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history_basic.history["accuracy"],
    label="Basic CNN Training"
)

plt.plot(
    history_basic.history["val_accuracy"],
    label="Basic CNN Validation"
)

plt.plot(
    history_alexnet.history["accuracy"],
    label="AlexNet Training"
)

plt.plot(
    history_alexnet.history["val_accuracy"],
    label="AlexNet Validation"
)

plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# STEP 12: VALIDATION ACCURACY COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history_basic.history["val_accuracy"],
    label="Basic CNN"
)

plt.plot(
    history_alexnet.history["val_accuracy"],
    label="AlexNet"
)

plt.title("Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# STEP 13: VALIDATION LOSS COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history_basic.history["val_loss"],
    label="Basic CNN"
)

plt.plot(
    history_alexnet.history["val_loss"],
    label="AlexNet"
)

plt.title("Validation Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# STEP 14: EVALUATION FUNCTION
# ============================================================

def evaluate_model(
    model,
    generator,
    model_name
):

    generator.reset()

    probabilities = model.predict(
        generator,
        verbose=1
    )

    predictions = (
        probabilities > 0.5
    ).astype(int).flatten()

    actual = generator.classes

    print("\n")
    print("=" * 60)
    print(model_name)
    print("=" * 60)

    print(
        classification_report(
            actual,
            predictions,
            target_names=[
                "Without Mask",
                "With Mask"
            ],
            digits=4
        )
    )

    cm = confusion_matrix(
        actual,
        predictions
    )

    plt.figure(figsize=(6, 5))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=[
            "Without Mask",
            "With Mask"
        ],
        yticklabels=[
            "Without Mask",
            "With Mask"
        ]
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.title(
        model_name +
        " - Confusion Matrix"
    )

    plt.show()

    return actual, predictions


# ============================================================
# STEP 15: EVALUATE BASIC CNN
# ============================================================

y_true_basic, y_pred_basic = evaluate_model(
    basic_cnn,
    val_gen,
    "Basic CNN"
)


# ============================================================
# STEP 16: EVALUATE ALEXNET
# ============================================================

y_true_alex, y_pred_alex = evaluate_model(
    alexnet,
    val_gen,
    "AlexNet"
)


# ============================================================
# STEP 17: PERFORMANCE COMPARISON TABLE
# ============================================================

basic_accuracy = (
    history_basic.history["val_accuracy"][-1]
    * 100
)

alexnet_accuracy = (
    history_alexnet.history["val_accuracy"][-1]
    * 100
)

comparison = pd.DataFrame(
    {
        "Model": [
            "Basic CNN",
            "AlexNet"
        ],

        "Validation Accuracy (%)": [
            round(basic_accuracy, 2),
            round(alexnet_accuracy, 2)
        ],

        "Total Parameters": [
            basic_cnn.count_params(),
            alexnet.count_params()
        ],

        "Training Time (seconds)": [
            round(time_basic, 2),
            round(time_alexnet, 2)
        ]
    }
)

print("\n==========================================")
print("MODEL PERFORMANCE COMPARISON")
print("==========================================")

print(
    comparison.to_string(
        index=False
    )
)


# ============================================================
# STEP 18: FINAL ACCURACY
# ============================================================

print("\n==========================================")
print("FINAL RESULTS")
print("==========================================")

print(
    f"Basic CNN Validation Accuracy : "
    f"{basic_accuracy:.2f}%"
)

print(
    f"AlexNet Validation Accuracy   : "
    f"{alexnet_accuracy:.2f}%"
)

print(
    f"Basic CNN Parameters           : "
    f"{basic_cnn.count_params():,}"
)

print(
    f"AlexNet Parameters             : "
    f"{alexnet.count_params():,}"
)

print(
    f"Basic CNN Training Time        : "
    f"{time_basic:.2f} seconds"
)

print(
    f"AlexNet Training Time          : "
    f"{time_alexnet:.2f} seconds"
)


# ============================================================
# STEP 19: SINGLE IMAGE PREDICTION
# ============================================================

def predict_mask(
    model,
    image_path,
    image_size=128
):

    # Read image
    image = cv2.imread(image_path)

    if image is None:
        print("Error: Image could not be loaded.")
        return

    # Convert BGR to RGB
    image_rgb = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # Resize
    resized = cv2.resize(
        image_rgb,
        (image_size, image_size)
    )

    # Normalize
    normalized = resized / 255.0

    # Add batch dimension
    input_image = np.expand_dims(
        normalized,
        axis=0
    )

    # Prediction
    probability = model.predict(
        input_image,
        verbose=0
    )[0][0]

    # IMPORTANT:
    # Check actual class indices
    class_indices = train_gen.class_indices

    if class_indices["with_mask"] == 1:

        if probability >= 0.5:
            label = "With Mask"
            confidence = probability
        else:
            label = "Without Mask"
            confidence = 1 - probability

    else:

        if probability >= 0.5:
            label = "Without Mask"
            confidence = probability
        else:
            label = "With Mask"
            confidence = 1 - probability

    # Display result
    plt.figure(figsize=(6, 6))

    plt.imshow(image_rgb)

    plt.title(
        f"Prediction: {label}\n"
        f"Confidence: {confidence * 100:.2f}%"
    )

    plt.axis("off")
    plt.show()

    print("Prediction :", label)
    print(
        "Confidence : "
        f"{confidence * 100:.2f}%"
    )


# ============================================================
# STEP 20: TEST WITH AN IMAGE WITHOUT MASK
# ============================================================

sample_without_mask = os.path.join(
    without_mask_dir,
    os.listdir(without_mask_dir)[0]
)

print("\nTesting image:")
print(sample_without_mask)

predict_mask(
    alexnet,
    sample_without_mask
)


# ============================================================
# STEP 21: TEST WITH AN IMAGE WITH MASK
# ============================================================

sample_with_mask = os.path.join(
    with_mask_dir,
    os.listdir(with_mask_dir)[0]
)

print("\nTesting image:")
print(sample_with_mask)

predict_mask(
    alexnet,
    sample_with_mask
)


# ============================================================
# EXPERIMENT COMPLETED
# ============================================================

print("\n")
print("=" * 60)
print("FACE MASK DETECTION EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 60)
```
